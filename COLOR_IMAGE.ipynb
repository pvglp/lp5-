```python
# Step 2: Import libraries
import os
import zipfile
import cv2
import shutil

from concurrent.futures import ThreadPoolExecutor
```

```python
# Step 3: Extract ZIP file

with zipfile.ZipFile("Files.zip", 'r') as zip_ref:
    zip_ref.extractall("Files")

print("✅ ZIP Extracted!")
```

```python
# Step 4: Create output folder

os.makedirs("output", exist_ok=True)
```

```python
# Step 5: Image processing function

def process_image(file):

    img = cv2.imread(f"Files/{file}")

    if img is None:
        return

    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Apply color map
    color = cv2.applyColorMap(gray, cv2.COLORMAP_JET)

    # Save image
    cv2.imwrite(f"output/color_{file}", color)

    print(f"Processed {file}")
```

```python
# Step 6: Get image files

files = [
    f for f in os.listdir("Files")
    if f.lower().endswith(('.jpg', '.png', '.jpeg'))
]

print(files)
```

```python
# Step 7: Parallel processing

with ThreadPoolExecutor(max_workers=4) as executor:
    executor.map(process_image, files)

print("✅ Processing Completed!")
```

```python
# Step 8: Create ZIP of output folder

shutil.make_archive("output_images", 'zip', "output")

print("✅ Output ZIP Created!")
```
